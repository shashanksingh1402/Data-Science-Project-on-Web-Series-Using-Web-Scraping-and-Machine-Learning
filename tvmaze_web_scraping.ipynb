```python
import requests
import pandas as pd
```

```python
all_data = []

for page in range(0, 5):   
    url = f"https://api.tvmaze.com/shows?page={page}"
    response = requests.get(url)
    
    shows = response.json()
    
    for show in shows:
        name = show.get("name")
        language = show.get("language")
        genres = ", ".join(show.get("genres", []))
        rating = show["rating"]["average"] if show["rating"] else None
        premiered = show.get("premiered")
        status = show.get("status")
        
        all_data.append([
            name, language, genres, rating, premiered, status
        ])
    
    print(f"Fetched page {page}")
```

```python
df = pd.DataFrame(
    all_data,
    columns=["Title", "Language", "Genres", "Rating", "Premiered", "Status"]
)

df.shape
```

```python
df.to_csv("web_series_raw_1200.csv", index=False)
```
